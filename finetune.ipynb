```python
import os
from datasets import load_dataset
from tqdm.notebook import tqdm

# 1. Output Directory Structure in Colab environment
output_dir = "/content/drive/MyDrive/pramila/dataset"
os.makedirs(os.path.join(output_dir, "images"), exist_ok=True)

# 2. Load Dataset from Hugging Face
print("Loading Teklia/IAM-line dataset from Hugging Face...")
ds = load_dataset("Teklia/IAM-line")

def prepare_paddle_data(split_name, txt_filename):
    txt_path = os.path.join(output_dir, txt_filename)
    split_data = ds[split_name]

    with open(txt_path, "w", encoding="utf-8") as f:
        for idx, sample in enumerate(tqdm(split_data, desc=f"Processing {split_name}")):
            image = sample["image"]
            text = sample["text"]

            # Clean ground-truth text
            text = text.strip().replace('"', '\\"')

            # Save image file
            img_filename = f"{split_name}_{idx:06d}.png"
            img_path = os.path.join(output_dir, "images", img_filename)

            # Ensure RGB image conversion
            if image.mode != "RGB":
                image = image.convert("RGB")
            image.save(img_path)

            # PaddleOCR label format: relative_path\ttext
            relative_img_path = os.path.join("images", img_filename)
            f.write(f"{relative_img_path}\t{text}\n")

# 3. Export splits
prepare_paddle_data("train", "train_list.txt")
prepare_paddle_data("validation", "val_list.txt")
prepare_paddle_data("test", "test_list.txt")

print(f"\nDataset successfully prepared in '{output_dir}'")
```

```python
import sys
import os
import cv2
import yaml

# 1. Set correct model_name for Detection and Recognition models
det_yml_path = "/content/drive/MyDrive/pramila/det_inference/inference.yml"
rec_yml_path = "/content/drive/MyDrive/pramila/inference/inference.yml"

def set_model_name(yml_path, model_name_str):
    if os.path.exists(yml_path):
        try:
            with open(yml_path, "r") as f:
                data = yaml.safe_load(f)
            if not isinstance(data, dict):
                data = {}
            if "Global" not in data or not isinstance(data["Global"], dict):
                data["Global"] = {}
            data["Global"]["model_name"] = model_name_str
            with open(yml_path, "w") as f:
                yaml.dump(data, f)
        except Exception as e:
            print(f"Warning updating {yml_path}: {e}")

set_model_name(det_yml_path, "PP-OCRv5_mobile_det")
set_model_name(rec_yml_path, "PP-OCRv5_mobile_rec")

# 2. Add PaddleOCR to python path
sys.path.append("/content/drive/MyDrive/pramila/PaddleOCR")

import tools.infer.utility as utility

# 3. Temporarily clear Jupyter kernel flags from sys.argv
sys.argv = ['']
args = utility.parse_args()

# 4. Assign custom paths and settings
args.det_model_dir = "/content/drive/MyDrive/pramila/det_inference"
args.rec_model_dir = "/content/drive/MyDrive/pramila/inference"
args.rec_char_dict_path = "/content/drive/MyDrive/pramila/inference/ppocr_keys.txt"
args.use_gpu = False
args.use_space_char = True
args.det_db_unclip_ratio = 2.0
args.use_angle_cls = False  # Set to False to fix the "cls model file path None" error
args.ir_optim = False
args.enable_mkldnn = False

from tools.infer.predict_system import TextSystem

# 5. Load custom models
text_sys = TextSystem(args)

# 6. Line-by-Line Re-sorting Function (Top-to-Bottom, Left-to-Right)
def sort_ocr_results(dt_boxes, rec_res, line_threshold=15):
    if dt_boxes is None or len(dt_boxes) == 0:
        return "No text detected."

    # Combine boxes and text
    results = []
    for box, (text, score) in zip(dt_boxes, rec_res):
        results.append([box.tolist(), (text, score)])

    # Sort top-to-bottom by Y-coordinate
    boxes = sorted(results, key=lambda b: b[0][0][1])

    lines = []
    current_line = []
    current_y = None

    for item in boxes:
        box, (text, score) = item[0], item[1]
        box_y = box[0][1]

        if current_y is None or abs(box_y - current_y) <= line_threshold:
            current_line.append(item)
            current_y = box_y
        else:
            current_line = sorted(current_line, key=lambda b: b[0][0][0])
            lines.append(current_line)
            current_line = [item]
            current_y = box_y

    if current_line:
        current_line = sorted(current_line, key=lambda b: b[0][0][0])
        lines.append(current_line)

    ordered_lines = []
    for line in lines:
        line_str = " ".join([item[1][0] for item in line])
        ordered_lines.append(line_str)

    return "\n".join(ordered_lines)


# 7. Run prediction on image
img_path = "/content/drive/MyDrive/pramila/dataset/js4.jpg"
img = cv2.imread(img_path)
dt_boxes, rec_res, _ = text_sys(img)

# 8. Output correctly sequenced text
ordered_text = sort_ocr_results(dt_boxes, rec_res, line_threshold=15)


print(ordered_text)
```

```python
import sys
import os
import cv2
import yaml
import logging

# 1. Silence PaddleOCR debug logs
logging.getLogger("ppocr").setLevel(logging.ERROR)
os.environ["FLAGS_use_mkldnn"] = "0"
os.environ["FLAGS_enable_pir_api"] = "0"

# 2. Change working directory to PaddleOCR
repo_path = "/content/drive/MyDrive/pramila/PaddleOCR"
os.chdir(repo_path)
sys.path.append(repo_path)

import ppstructure.utility as utility
from ppstructure.predict_system import StructureSystem

# 3. Set model_name in inference.yml
det_yml_path = "/content/drive/MyDrive/pramila/det_inference/inference.yml"
rec_yml_path = "/content/drive/MyDrive/pramila/inference/inference.yml"

def set_model_name(yml_path, model_name_str):
    if os.path.exists(yml_path):
        try:
            with open(yml_path, "r") as f: data = yaml.safe_load(f)
            if not isinstance(data, dict): data = {}
            if "Global" not in data or not isinstance(data["Global"], dict): data["Global"] = {}
            data["Global"]["model_name"] = model_name_str
            with open(yml_path, "w") as f: yaml.dump(data, f)
        except Exception: pass

set_model_name(det_yml_path, "PP-OCRv5_mobile_det")
set_model_name(rec_yml_path, "PP-OCRv5_mobile_rec")

# 4. Setup arguments
sys.argv = ['']
args = utility.parse_args()

args.det_model_dir = "/content/drive/MyDrive/pramila/det_inference"
args.rec_model_dir = "/content/drive/MyDrive/pramila/inference"
args.rec_char_dict_path = "/content/drive/MyDrive/pramila/inference/ppocr_keys.txt"
args.layout_model_dir = "/content/drive/MyDrive/pramila/picodet_lcnet_x1_0_layout_infer"
args.layout_dict_path = os.path.join(repo_path, "ppocr/utils/dict/layout_dict/layout_publaynet_dict.txt")

args.table_char_dict_path = os.path.join(repo_path, "ppocr/utils/dict/table_structure_dict_ch.txt")
args.table_structure_dict_path = os.path.join(repo_path, "ppocr/utils/dict/table_structure_dict_ch.txt")
args.table = False

args.use_gpu = False
args.enable_mkldnn = False
args.ir_optim = False
args.use_space_char = True
args.use_angle_cls = False
args.show_log = False  # Turn off internal debug logging

# 5. Initialize StructureSystem engine
engine = StructureSystem(args)

# 6. Extractor Function that extracts ALL detected text regions
def extract_universal_document(img_path):
    img = cv2.imread(img_path)
    if img is None:
        return "Error: Image not found at specified path."

    result, _ = engine(img)
    extracted_output = []

    for block in result:
        res = block.get('res', [])

        # Extract text lines inside the block
        if isinstance(res, list) and len(res) > 0:
            lines = []
            for item in res:
                if isinstance(item, dict) and 'text' in item:
                    lines.append(item['text'])
            if lines:
                extracted_output.append(" ".join(lines))
        elif isinstance(res, dict) and 'html' in res:
            extracted_output.append(f"\n[TABLE]\n{res['html']}\n")

    return "\n\n".join(extracted_output)

# 7. Run inference
image_path = "/content/drive/MyDrive/pramila/dataset/js4.jpg"
print("\n================ EXTRACTED TEXT ================")
print(extract_universal_document(image_path))
print("================================================")
```
